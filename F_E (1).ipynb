### Feature Encoding

```python
# converting categorical data into numerical data 
# popular method for feature encoding:
1.label Encoding : when their is no order between the categories.
2.Ordinal Encoding: when their is order between the categories.
3.One hot Encoding: when number of entries is less in categorical column.
```

```python
# into numerical dataa 
from sklearn.preprocessing import LabelEncoder
import pandas as pd
data = {'Color': ['White','Red','Yellow','Blue','Green']}
df = pd.DataFrame(data)
encoder = LabelEncoder()# create the labelencoder
encoder.fit(df['Color'])# fit() allot the interger value to each category
# Transform() transform the actual data based upon fit() function allocation
df['encoded_color'] = encoder.transform(df['Color'])
print(df)
```

# 2.Ordinal Encoding

```python
# convert categorical data into numerical data based upon the raking.
from sklearn.preprocessing import OrdinalEncoder
data =[['phd'],['PG'],['UG'],['UG'],['UG'],['phd']]
encoder = OrdinalEncoder(categories =[['UG','PG','phd']])
encoder_data = encoder.fit_transform(data)
print(encoder_data)
```

```python
# Use label encoder when we don't have raking in the categories.
# use  ordinal encoder when we have raking in the categories.
```

# 3. One Hot Encoder

```python
# convert categorical data into binary indicators.
import pandas as pd
data ={"country":['India','USA','Uganda','Japan','Thailand','Nigeria','India']}
df = pd.DataFrame(data)
encoded_data = pd.get_dummies(df,dtype = int)
encoded_data
```

# Feature scaling : to bring the data in particular range.

```python
# Ml models give more important to large values.
# so we bring all the data same range(0 to 1)
# Types of Feature scaling 
1.min max scaling
2.standardization (z-score scaling)
3.Robust scaling
```

# Min Max Scaling

```python
# when we use is when they is no outliers.
# Min max scaler bring the value by deafult in 0 to 1 range.
# x_scaled = (x-x_min) / (x_max - x_min)
# x =[100000,200000,300000,400000,500000]
# x = 100000 => 0
# x = 200000 => 0.25
# x = 300000 => 0.5
# x = 400000 => 0.75
# x = 500000 => 1
```

```python
from sklearn.preprocessing import MinMaxScaler
import pandas as pd
scaler = MinMaxScaler()
data = [100000,200000,300000,400000,500000]
df = pd.DataFrame(data,columns = ["values"])
df['scaled_values'] = scaler.fit_transform(df[['values']])
df
# minmaxscaler cannot handle the outliers
# when the one value is large 
```

# Robust Scaler 

```python
# use  the robust scaler when their is outliers in the dataset 
# Robust scaler is based upon IQR and median
# X_scaled = (X- Median(X))/IQR(X)
# IQR : inter quartile range = Q3 - Q1
```

```python
from sklearn.preprocessing import RobustScaler
scaler = RobustScaler()
data = [100,200,300,400,5000]
df = pd.DataFrame(data,columns = ["values"])
df['scaled_value']=scaler.fit_transform(df[['values']])
df
```

```python
# manually 
median = np.median(df.values)
Q1 = np.percentile(df.values,25)
Q3 = np.percentile(df.values,75)
print(Q1,Q3)
IQR = Q3 - Q1
X = 100
X_scaled = (X-median)/IQR
X_scaled
```
